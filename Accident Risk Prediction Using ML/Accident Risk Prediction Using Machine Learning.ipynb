  Introduction
</div>

<h3 style="font-weight: bold; color: #032B44;">Problem Statement :</h3>

Road accidents are one of the major causes of death and injury worldwide, often resulting from a combination of driver behavior, adverse weather conditions, road surface issues, and high traffic density. Predicting the likelihood of an accident before it occurs can help authorities take preventive actions and ensure safer transportation.

This project aims to build a machine learning model that predicts whether a vehicle accident may occur, based on historical accident data including driver details, road conditions, traffic information, and weather conditions.

<h3 style="font-weight: bold; color: #032B44;">About the Dataset :</h3>

The dataset used in this project is a vehicle road accident dataset containing accident-related records collected from real-world traffic observations. It includes various features related to driver details, road conditions, traffic factors, and weather conditions that influence the probability of an accident occurring.

<h3 style="font-weight: bold; color: #032B44;"> Overview :</h3>

The project aim to build a machine learning model that predict whether a vehicle accident may occur , based on historical accident data including driver details , road conditions , traffic information , and weather conditions.

**Build a machine learning model that predict whether a vehicle accident may occur.**

<h3 style="font-weight: bold; color: #032B44;">Weather Conditions vs Number of Accidents</h3>

<h3 style="font-weight: bold; color: #032B44;">Road Condition vs Accident Severity</h3>

```python
plt.figure(figsize=(8,5))

sns.countplot(data=gra_data, x="Road Condition", hue="Accident Severity", palette=['#032B44', '#4682B4', '#2E865F'])

plt.title("Road Condition vs Accident Severity")
plt.xlabel("Road Condition")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.show()
```

* Wet and Icy roads have the highest overall accident counts, with Severe accidents slightly outnumbering Moderate and Minor ones.
* Snow‑covered roads show a balanced distribution among Moderate, Minor, and Severe accidents, all near the 11,000 mark.

**Reason 1 — Insufficient Feature Representation**

* Some important accident-related factors (like time of day, road quality, exact driver distraction reason, visibility range, traffic density) may not be included in the dataset, causing confusion between similar cases.
* For example, two accidents may have similar driver age and weather conditions but differ in unseen environmental conditions, leading to misclassification.

**Reason 2 — Class Imbalance**

* If one class (e.g., Minor accidents) appears much more frequently than others (Severe or Fatal), the model may predict the majority class more often.

**Most misclassifications occurred between Minor and Major accident severity levels. A key reason is that driver age, vehicle type, and weather conditions overlap heavily between these groups, making class boundaries difficult to separate. Additionally, the dataset has class imbalance, where minor accidents dominate the data, which biases the model toward predicting the majority class.**
