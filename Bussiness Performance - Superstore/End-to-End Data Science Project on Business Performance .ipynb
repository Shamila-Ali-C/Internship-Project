  Data Cleaning and preprocessing
</div>

```python
quality = pd.DataFrame({
    "Missing Values": df.isna().sum(),
    "Missing %": df.isna().mean() * 100,
    "Unique Values": df.nunique(),
    "Data Type": df.dtypes.astype(str)
}).sort_values("Missing Values", ascending=False)

display(quality)
```

```python
# Check invalid numeric values
print('Negative Sales:', (df['Sales'] < 0).sum())
print('Negative Quantity:', (df['Quantity'] < 0).sum())
print('Discount outside 0-1:', ((df['Discount'] < 0) | (df['Discount'] > 1)).sum())
```

```python
# Create a working copy
data = df.copy()
```

```python
# Standardize column names
data.columns = (
    data.columns
    .str.strip()
    .str.replace(' ', '_')
    .str.replace('-', '_')
)
```

<div style="color: white; background-color: #4682B4; font-size: 28px; font-weight: bold; text-align: center; padding: 10px 20px; border-radius: 20px;">
  Feature Engineering
</div>

```python
# Convert date columns
data['Order_Date'] = pd.to_datetime(data['Order_Date'], errors='coerce')
data['Ship_Date'] = pd.to_datetime(data['Ship_Date'], errors='coerce')
```

```python
# Create shipping duration
data['Shipping_Days'] = (data['Ship_Date'] - data['Order_Date']).dt.days
```

```python
# Add business-friendly time dimensions
data['Year'] = data['Order_Date'].dt.year
data['Month'] = data['Order_Date'].dt.month
data['Month_Name'] = data['Order_Date'].dt.strftime('%b')
data['Year_Month'] = data['Order_Date'].dt.to_period('M').astype(str)
data['Quarter'] = data['Order_Date'].dt.quarter
```

```python
# Profit margin at transaction level
data['Profit_Margin'] = np.where(
    data['Sales'] != 0,
    data['Profit'] / data['Sales'] * 100,
    0
)
```

```python
print("\nColumn names:")
for i, col in enumerate(data.columns, start=1):
    print(f"{i:02d}. {col}")
```

<div style="color: white; background-color: #4682B4; font-size: 28px; font-weight: bold; text-align: center; padding: 10px 20px; border-radius: 20px;">
  Business KPIs (Key Performance Indicator) Analysis
</div>

```python
total_sales = data['Sales'].sum()
total_profit = data['Profit'].sum()
total_quantity = data['Quantity'].sum()
total_orders = data['Order_ID'].nunique()
total_customers = data['Customer_ID'].nunique()
avg_order_value = data.groupby('Order_ID')['Sales'].sum().mean()
profit_margin = total_profit / total_sales * 100
```

```python
kpis = pd.DataFrame({
    'KPI': [
        'Total Sales', 'Total Profit', 'Total Quantity',
        'Total Orders', 'Unique Customers',
        'Average Order Value', 'Overall Profit Margin'
    ],
    'Value': [
        total_sales, total_profit, total_quantity,
        total_orders, total_customers,
        avg_order_value, profit_margin
    ]
})
display(kpis)
```

<div style="color:#1B4F72;font-size: 24px;font-weight: bold; text-align: left; font-family: 'Times New Roman', serif; margin-top: 10px; margin-bottom: 15px;">
  Sales and Profit by Category
</div>

```python
category_perf = (
    data.groupby("Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order_ID", "nunique")
    )
)
category_perf["Profit_Margin_%"] = category_perf["Profit"] / category_perf["Sales"] * 100
display(category_perf.sort_values("Sales", ascending=False))
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
category_perf[['Sales', 'Profit']].plot(kind='bar', ax=ax, color=['#0B1F3A', 'blue'])
ax.set_title('Sales and Profit by Category')
ax.set_ylabel('Amount')
ax.tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()
```

<div style="color: white; background-color: #4682B4; font-size: 28px; font-weight: bold; text-align: center; padding: 10px 20px; border-radius: 20px;">
  Sub-Category Analysis
</div>

```python
subcategory_perf = (
    data.groupby('Sub_Category')
    .agg(
        Sales=('Sales', 'sum'),
        Profit=('Profit', 'sum'),
        Quantity=('Quantity', 'sum')
    )
)

subcategory_perf['Profit_Margin_%'] = (
    subcategory_perf['Profit'] / subcategory_perf['Sales'] * 100
)

display(subcategory_perf.sort_values('Profit', ascending=False))
```

```python
fig, ax = plt.subplots(figsize=(12, 7))
plot_data = subcategory_perf.sort_values('Profit')
plot_data['Profit'].plot(kind='barh', ax=ax, color='#0B1F3A')
ax.set_title('Profit by Sub-Category')
ax.set_xlabel('Profit')
plt.tight_layout()
plt.show()
```

**Category analysis separates revenue leadership from profitability leadership. A category can generate strong sales while producing weaker margins. Sub-category analysis is therefore essential for identifying specific product groups that require pricing, discount, or cost review.**

**Time-series analysis identifies growth patterns, seasonality, and changes in business performance across years and months.**

<div style="color:#1B4F72;font-size: 24px;font-weight: bold; text-align: left; font-family: 'Times New Roman', serif; margin-top: 10px; margin-bottom: 15px;">
  Yearly Performance
</div>

```python
yearly_perf = (
    data.groupby("Year")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum")
    )
)
yearly_perf["Sales_Growth_%"] = yearly_perf["Sales"].pct_change() * 100
yearly_perf["Profit_Growth_%"] = yearly_perf["Profit"].pct_change() * 100
yearly_perf["Profit_Margin_%"] = yearly_perf["Profit"] / yearly_perf["Sales"] * 100

display(yearly_perf)

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(yearly_perf.index, yearly_perf["Sales"], marker="o", linewidth=2, label="Sales", color='blue')
ax.plot(yearly_perf.index, yearly_perf["Profit"], marker="o", linewidth=2, label="Profit", color='red')
ax.set_title("Yearly Sales and Profit Trend")
ax.set_xlabel("Year")
ax.set_ylabel("Amount")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
yearly_perf = (
    data.groupby('Year')
    .agg(
        Sales=('Sales', 'sum'),
        Profit=('Profit', 'sum'),
        Quantity=('Quantity', 'sum')
    )
)

yearly_perf['Sales_Growth_%'] = yearly_perf['Sales'].pct_change() * 100
yearly_perf['Profit_Growth_%'] = yearly_perf['Profit'].pct_change() * 100
yearly_perf['Profit_Margin_%'] = yearly_perf['Profit'] / yearly_perf['Sales'] * 100

display(yearly_perf)

fig, ax = plt.subplots(figsize=(11, 5))
yearly_perf[['Sales', 'Profit']].plot(kind='bar', ax=ax, color= ['#0B1F3A', 'blue'])
ax.set_title('Yearly Sales and Profit')
ax.set_ylabel('Amount')
ax.tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()
```
