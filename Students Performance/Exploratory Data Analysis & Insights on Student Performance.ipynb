  Abstract
</div>

**This project focuses on performing Exploratory Data Analysis (EDA) to identify meaningful patterns in student performance data. The UCI Student Performance dataset for Mathematics and Portuguese courses is analyzed using demographic, family, educational, lifestyle, attendance, and academic variables, including G1, G2, and G3 grades.**

**The datasets are cleaned and analyzed through univariate, bivariate, and multivariate techniques using Python, Pandas, NumPy, Matplotlib, and Seaborn. The analysis shows that Portuguese students have higher average final grades in the supplied sample, previous failures are negatively associated with performance, study time has a modest positive association, and absences have a weak relationship with final grades. Earlier grades, particularly G2, show the strongest association with final performance. Overall, the project demonstrates how EDA can transform educational data into meaningful and interpretable insights.**

  Data Quality Checks
</div>

```python
def quality_report(df, name):
    report = pd.DataFrame({
        'dtype': df.dtypes,
        'missing_values': df.isna().sum(),
        'missing_%': df.isna().mean() * 100,
        'unique_values': df.nunique(),
        'duplicate_rows': df.duplicated().sum()
    })
    print(f'{name}: {df.shape[0]} rows, {df.shape[1]} columns')
    print('Total missing values:', df.isna().sum().sum())
    print('Duplicate rows:', df.duplicated().sum())
    return report

math_quality = quality_report(math_df, 'Mathematics')
port_quality = quality_report(por_df, 'Portuguese')

display(math_quality)
```

<div style="color: white; background-color: #0B1F3A; font-size: 28px; font-weight: bold; text-align: center; padding: 10px 20px; border-radius: 20px;">
  Matched Students Profile Analysis
</div>

**There is no explicit unique student ID in the supplied files. Therefore, we match using the common student-background variables and exclude the subject-specific performance variables (`G1`, `G2`, `G3`, `absences`). This produces a subset of students that can be identified consistently across the two files.**

**Because matching is based on available attributes rather than a true ID, this analysis should be described as a `matched-profile analysis`, not as a guaranteed longitudinal student identifier.**

```python
match_cols = [c for c in math_df.columns if c not in ['G1','G2','G3','absences']]
matched = pd.merge(math_df, por_df, on=match_cols, how='inner', suffixes=('_math','_port'))
print('Matched rows:', len(matched))
print('Matched columns:', matched.shape[1])
display(matched[['G3_math','G3_port','absences_math','absences_port']].head())
```

```python
fig, ax = plt.subplots(figsize=(7,5))
sns.scatterplot(data=matched, x='G3_math', y='G3_port', alpha=.7, ax=ax)
sns.regplot(data=matched, x='G3_math', y='G3_port', scatter=False, ax=ax)
ax.set_title('Matched Students: Mathematics vs Portuguese Final Grade')
ax.set_xlabel('Math G3')
ax.set_ylabel('Portuguese G3')
plt.show()

print('Correlation between Math G3 and Portuguese G3:', matched[['G3_math','G3_port']].corr().iloc[0,1])
print('Mean Math G3:', matched.G3_math.mean())
print('Mean Portuguese G3:', matched.G3_port.mean())
```
